```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/dataset.csv")
print("Shape:", df.shape)
df.head()
```

```python
y = df["Class"].values        
X_raw = df.drop(columns=["Class"])

print("X shape:", X_raw.shape)
print("y shape:", y.shape)
print("Broj klasa:", np.unique(y).size)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=30, random_state=42)
X_pca = pca.fit_transform(X_scaled)
```

```python
from sklearn.mixture import GaussianMixture

K = np.unique(y).size

gmm = GaussianMixture(
    n_components=K,
    covariance_type="full",   # najopštiji i najrealniji
    n_init=10,
    random_state=42
)

clusters = gmm.fit_predict(X_pca)

np.unique(clusters, return_counts=True)
```

```python
pd.DataFrame(
    clusters, columns=["cluster"]
).to_csv(
    "../results/gmm_clusters_full.csv", index=False
)
```

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    adjusted_rand_score,
    normalized_mutual_info_score
)

sil = silhouette_score(X_pca, clusters)
db = davies_bouldin_score(X_pca, clusters)
ari = adjusted_rand_score(y, clusters)
nmi = normalized_mutual_info_score(y, clusters)

sil, db, ari, nmi
```

```python
import matplotlib.pyplot as plt

pca_2d = PCA(n_components=2, random_state=42)
X_2d = pca_2d.fit_transform(X_pca)

plt.figure(figsize=(7,5))
plt.scatter(
    X_2d[:,0],
    X_2d[:,1],
    c=clusters,
    s=5,
    cmap="tab10"
)
plt.title("GMM clustering (PCA 2D)")
plt.tight_layout()
plt.show()
```

```python
results = pd.DataFrame([{
    "Algorithm": "GMM",
    "K": K,
    "Silhouette": sil,
    "Davies_Bouldin": db,
    "ARI": ari,
    "NMI": nmi
}])

results.to_csv("../results/gmm_results.csv", index=False)
results
```
