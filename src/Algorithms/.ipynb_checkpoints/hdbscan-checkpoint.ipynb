```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import hdbscan
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score
from sklearn.model_selection import ParameterGrid
from sklearn.manifold import TSNE
import umap

datasets = {
    "full": "../../data/processed/ProcessedData.csv",
    "pca": "../../data/reduced/dataPCA.csv",
    "fs": "../../data/reduced/dataFS.csv"
}

outDir = "../../results/hdbscan"
```

```python
def plotHDB(X, labels, method, name):
    plt.figure(figsize=(6,5))
    noise = labels == -1
    plt.scatter(X[noise,0], X[noise,1], c="lightgray", s=5, label="Noise")
    plt.scatter(X[~noise,0], X[~noise,1], c=labels[~noise], cmap="tab10", s=5)
    plt.title(f"HDBSCAN {method} ({name})")
    plt.tight_layout()
    plt.savefig(f"{outDir}/hdbscan_{method}_{name}.png", dpi=200)
    plt.close()
```

```python
param_grid = {
    "min_cluster_size": [5, 10, 15, 30],
    "min_samples": [5, 10, 20]
}

results = []

for name, path in datasets.items():
    
    df = pd.read_csv(path)

    if "Class" in df.columns:
        y_true = df["Class"].values
        X = df.drop(columns=["Class"])
    else:
        y_true = None
        X = df.copy()

    if "V129" in X.columns:
        X = X.drop(columns=["V129"])

    X_np = X.values

    best_score = -1
    best_params = None
    best_labels = None

    for params in ParameterGrid(param_grid):
        model = hdbscan.HDBSCAN(**params, metric="euclidean")
        labels = model.fit_predict(X_np)

        clusters = set(labels) - {-1}
        if len(clusters) < 2:
            continue

        mask = labels != -1
        sil = silhouette_score(X_np[mask], labels[mask])

        if sil > best_score:
            best_score = sil
            best_params = params
            best_labels = labels

    labels = best_labels

    clusters = set(labels) - {-1}
    num_clusters = len(clusters)
    noise_ratio = np.mean(labels == -1)

    mask = labels != -1
    if len(set(labels[mask])) > 1:
        sil = silhouette_score(X_np[mask], labels[mask])
        db  = davies_bouldin_score(X_np[mask], labels[mask])
    else:
        sil, db = np.nan, np.nan

    if y_true is not None:
        ari = adjusted_rand_score(y_true, labels)
        nmi = normalized_mutual_info_score(y_true, labels)
    else:
        ari, nmi = np.nan, np.nan

    out = X.copy()
    out["cluster"] = labels
    out.to_csv(f"{outDir}/hdbscan_{name}_clusters.csv", index=False)

    reducer = umap.UMAP(n_components=2, random_state=42)
    X_umap = reducer.fit_transform(X_np)
    plotHDB(X_umap, labels, "UMAP", name)

    tsne = TSNE(n_components=2, perplexity=30, random_state=42)
    X_tsne = tsne.fit_transform(X_np)
    plotHDB(X_tsne, labels, "TSNE", name)

    results.append({
        "Dataset": name,
        "Algorithm": "HDBSCAN",
        "min_cluster_size": best_params["min_cluster_size"],
        "min_samples": best_params["min_samples"],
        "Clusters": num_clusters,
        "NoiseRatio": noise_ratio,
        "Silhouette": sil,
        "Davies_Bouldin": db,
        "ARI": ari,
        "NMI": nmi
    })
```

```python
summaryData = pd.DataFrame(results)
summaryData.to_csv(f"{outDir}/hdbscan_summary.csv", index=False)
print(summaryData)
```
